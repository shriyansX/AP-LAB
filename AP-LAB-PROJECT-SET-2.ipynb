```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Read monthly sales data from CSV file
data = pd.read_csv("sales_data.csv")

print("Sales Data:")
print(data)

# Extract columns
months = data['Month']
sales = data['Sales']
advertising = data['Advertising']

# 2. Calculate total and average sales
total_sales = sales.sum()
average_sales = sales.mean()

print("\nTotal Sales:", total_sales)
print("Average Sales:", average_sales)

# 3. Identify highest and lowest sales month
highest_month = data.loc[sales.idxmax()]
lowest_month = data.loc[sales.idxmin()]

print("\nHighest Sales Month:")
print(highest_month)

print("\nLowest Sales Month:")
print(lowest_month)

# 4. Filter sales greater than threshold
threshold = 2000
high_sales = data[data['Sales'] > threshold]

print("\nSales greater than", threshold)
print(high_sales)

# 5. Convert sales list into NumPy array and compute statistics
sales_array = np.array(sales)

print("\nNumPy Statistics")
print("Mean:", np.mean(sales_array))
print("Standard Deviation:", np.std(sales_array))
print("Max:", np.max(sales_array))
print("Min:", np.min(sales_array))

# 6. Bar chart for product sales comparison
plt.figure()
plt.bar(months, sales)
plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()

# 7. Line graph for monthly sales trend
plt.figure()
plt.plot(months, sales, marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()

# 8. Correlation between advertising cost and sales
correlation = np.corrcoef(advertising, sales)

print("\nCorrelation Matrix:")
print(correlation)

# 9. Matrix multiplication example (sales forecasting)
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

result = np.dot(A,B)

print("\nMatrix Multiplication Result:")
print(result)
```
