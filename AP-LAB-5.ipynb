```python
student = {
    "Name": "Rahul",
    "Roll No": 101,
    "Course": "B.Tech",
    "Branch": "CSE",
    "Year": 2
}

print("Keys:")
for key in student:
    print(key)

print("\nValues:")
for value in student.values():
    print(value)
```

```python
sentence = input("Enter a sentence: ")

words = sentence.lower().split()

freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print("Word Frequency:")
for word, count in freq.items():
    print(word, ":", count)
```

```python
dict1 = {
    "a": 1,
    "b": 2,
    "c": 3
}

dict2 = {
    "d": 4,
    "e": 5
}

merged_dict = {}

for key in dict1:
    merged_dict[key] = dict1[key]

for key in dict2:
    merged_dict[key] = dict2[key]
    
print("Merged Dictionary:")
print(merged_dict)
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7}

union_set = set1 | set2

intersection_set = set1 & set2

difference_set = set1 - set2

print("Set 1:", set1)
print("Set 2:", set2)
print("Union:", union_set)
print("Intersection:", intersection_set)
print("Difference (Set1 - Set2):", difference_set)
```

```python
my_list = [1, 2, 3, 4, 2, 3, 5, 1]

unique_list = list(set(my_list))

print("Original List:", my_list)
print("List after removing duplicates:", unique_list)
```

```python
data = {
    "banana": 3,
    "apple": 5,
    "cherry": 1,
    "mango": 4
}

sorted_by_keys = dict(sorted(data.items()))

sorted_by_values = dict(sorted(data.items(), key=lambda item: item[1]))

print("Original Dictionary:", data)
print("Sorted by Keys:", sorted_by_keys)
print("Sorted by Values:", sorted_by_values)
```

```python
data = {
    "a": 1,
    "b": 2,
    "c": 3
}

tuple_list = list(data.items())

new_list = [("x", 10), ("y", 20), ("z", 30)]

new_dict = dict(new_list)

print("Original Dictionary:", data)
print("Dictionary to List of Tuples:", tuple_list)
print("List of Tuples:", new_list)
print("List of Tuples to Dictionary:", new_dict)
```

```python
college = {
    "CSE": {
        "Rahul": 85,
        "Anita": 92,
        "Soham": 78
    },
    "ECE": {
        "Amit": 88,
        "Neha": 90
    },
    "ME": {
        "Rohan": 75,
        "Priya": 82
    }
}

for department, students in college.items():
    print("Department:", department)
    for student, marks in students.items():
        print("  Student:", student, "| Marks:", marks)
```

```python
phone_directory = {}

while True:
    print("\n--- Phone Directory Menu ---")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Update Contact")
    print("4. Delete Contact")
    print("5. Display All Contacts")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        name = input("Enter name: ")
        phone = input("Enter phone number: ")
        phone_directory[name] = phone
        print("Contact added successfully.")

    elif choice == 2:
        name = input("Enter name to search: ")
        if name in phone_directory:
            print("Phone Number:", phone_directory[name])
        else:
            print("Contact not found.")

    elif choice == 3:
        name = input("Enter name to update: ")
        if name in phone_directory:
            phone = input("Enter new phone number: ")
            phone_directory[name] = phone
            print("Contact updated successfully.")
        else:
            print("Contact not found.")

    elif choice == 4:
        name = input("Enter name to delete: ")
        if name in phone_directory:
            del phone_directory[name]
            print("Contact deleted successfully.")
        else:
            print("Contact not found.")

    elif choice == 5:
        if phone_directory:
            print("Phone Directory:")
            for name, phone in phone_directory.items():
                print(name, ":", phone)
        else:
            print("Directory is empty.")

    elif choice == 6:
        print("Exiting Phone Directory.")
        break

    else:
        print("Invalid choice. Try again.")
```
