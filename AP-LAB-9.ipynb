```python
import pandas as pd

df = pd.read_csv("students.csv")

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
```

```python
filtered = df[df["Marks"] > 60]

print(filtered)
```

```python
import numpy as np

# list
numbers = [10, 20, 30, 40, 50]

# convert list to numpy array
arr = np.array(numbers)

print("Array:", arr)

# operations
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Square:", arr**2)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# read csv file
df = pd.read_csv("students.csv")

# bar chart
plt.bar(df["Name"], df["Marks"])

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks Bar Chart")

plt.show()
```

```python
import matplotlib.pyplot as plt

# given data
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 40]

# line graph
plt.plot(x, y)

plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Line Graph")

plt.show()
```

```python
import pandas as pd

data = {
    "Maths": [80, 70, 90, 60],
    "Science": [85, 75, 95, 65],
    "English": [78, 72, 88, 58]
}

df = pd.DataFrame(data)

# correlation matrix
corr_matrix = df.corr()

print(corr_matrix)
```

```python
import numpy as np

# matrix A
A = np.array([[1, 2],
              [3, 4]])

# matrix B
B = np.array([[5, 6],
              [7, 8]])

# matrix multiplication
result = np.dot(A, B)

print(result)
```
