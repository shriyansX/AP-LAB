## 1. Read a Text File and Count Characters

```python
try:
    with open("sample.txt", "r") as file:
        content = file.read()
        print("Number of characters:", len(content))
except FileNotFoundError:
    print("File not found.")
```

## 2. Append Data to a File

```python
data = input("Enter text to append: ")

with open("sample.txt", "a") as file:
    file.write("\n" + data)

print("Data appended successfully.")
```

## 3. Copy Content from One File to Another

```python
try:
    with open("source.txt", "r") as src:
        content = src.read()

    with open("destination.txt", "w") as dest:
        dest.write(content)

    print("File copied successfully.")
except FileNotFoundError:
    print("Source file not found.")
```

## 4. Count Lines, Words, and Characters

```python
try:
    with open("sample.txt", "r") as file:
        lines = file.readlines()

        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)

    print("Lines:", line_count)
    print("Words:", word_count)
    print("Characters:", char_count)
except FileNotFoundError:
    print("File not found.")
```

## 5. Read a File and Remove Punctuation

```python
import string

try:
    with open("sample.txt", "r") as file:
        content = file.read()

    no_punctuation = content.translate(str.maketrans('', '', string.punctuation))

    print("Text without punctuation:\n")
    print(no_punctuation)
except FileNotFoundError:
    print("File not found.")
```

## 6. Create a Log File and Write Error Messages

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print("Result:", result)

except Exception as e:
    with open("error.log", "a") as log:
        log.write(str(e) + "\n")
    print("Error logged successfully.")
```

## 7. Store Student Records in a CSV File

```python
import csv

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Name", "Roll", "Marks"])

    for i in range(3):
        name = input("Enter name: ")
        roll = input("Enter roll: ")
        marks = input("Enter marks: ")
        writer.writerow([name, roll, marks])

print("Student records saved in CSV file.")
```

## 8. Read a CSV Using Pandas and Print First 5 Rows
