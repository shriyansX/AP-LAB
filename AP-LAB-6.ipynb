```python
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

print(factorial(5))
```

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(7))
```

```python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

```python
def list_stats(lst):
    total = sum(lst)
    avg = total / len(lst)
    maximum = max(lst)
    minimum = min(lst)
    return total, avg, maximum, minimum

result = list_stats([10, 20, 30, 40])
print(result)
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(6))
```

```python
words = ["python", "java", "c"]
upper_words = list(map(str.upper, words))

print(upper_words)
```

```python
from functools import reduce

numbers = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, numbers)

print(product)
```

```python
def greet(name="Student", message="Welcome"):
    print(message, name)

greet()
greet(name="Shriyans")
greet(message="Hello", name="Shriyans")
```

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)

print(fib(10))
```
